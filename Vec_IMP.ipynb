Full Rank vs Rank Deficiency

For a matrix with 3 columns:

Rank = 3 → full rank
Rank < 3 → rank deficient

Rank deficiency means some columns/features contain redundant information.

```python
A = np.array([[1,2],
              [3,4]])

print(np.linalg.matrix_rank(A))
```

```python
import numpy as np
import matplotlib.pyplot as plt

v = np.array([3, 2])

plt.quiver(
    0, 0,          # starting point
    v[0], v[1],    # vector components
    angles='xy',
    scale_units='xy',
    scale=1
)

plt.xlim(-1, 5)
plt.ylim(-1, 4)

plt.axhline(0)
plt.axvline(0)

plt.grid()
plt.xlabel("X")
plt.ylabel("Y")
plt.title("2D Vector")

plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

a = np.array([3, 2])
b = np.array([2, 1])

addition = a + b
subtraction = a - b

print("a + b =", addition)
print("a - b =", subtraction)

plt.quiver(0, 0, a[0], a[1],
           angles='xy', scale_units='xy', scale=1, label='a')

plt.quiver(0, 0, b[0], b[1],
           angles='xy', scale_units='xy', scale=1, label='b')

plt.quiver(0, 0, addition[0], addition[1],
           angles='xy', scale_units='xy', scale=1, label='a + b')

plt.xlim(-1, 7)
plt.ylim(-1, 5)
plt.axhline(0)
plt.axvline(0)
plt.grid()
plt.legend()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

a = np.array([3, 2])
b = np.array([-2,3])

dot_product = np.dot(a, b)

print("Dot product:", dot_product)


plt.quiver(0,
           0, a[0], a[1],
           angles='xy', scale_units='xy', scale=1, label='a')

plt.quiver(0, 0, b[0], b[1],
           angles='xy', scale_units='xy', scale=1, label='b')

plt.xlim(-1, 4)
plt.ylim(-1, 4)

plt.axhline(0)
plt.axvline(0)
plt.grid()
plt.legend()

plt.title(f"Dot Product = {dot_product}")
plt.show()
```

```python
v = np.array([3, 4])

magnitude = np.linalg.norm(v)

normalized = v / magnitude

print("Original:", v)
print("Magnitude:", magnitude)
print("Normalized:", normalized)
print("New magnitude:", np.linalg.norm(normalized))

plt.quiver(0, 0, v[0], v[1],
           angles='xy', scale_units='xy', scale=1, label='Original')

plt.quiver(0, 0, normalized[0], normalized[1],
           angles='xy', scale_units='xy', scale=1, color="red",label='Normalized')

plt.xlim(-1, 4)
plt.ylim(-1, 5)

plt.axhline(0)
plt.axvline(0)
plt.grid()
plt.legend()

plt.title("Vector Normalization")
plt.show()
```

```python
#euclidean and manhattan

import numpy as np
import matplotlib.pyplot as plt

A = np.array([1, 1])
B = np.array([4, 3])

euclidean = np.linalg.norm(B - A)
manhattan = np.linalg.norm(B - A, ord=1)

print("Euclidean distance:", euclidean)
print("Manhattan distance:", manhattan)

plt.figure(figsize=(7, 7))

# Points
plt.scatter(*A, s=100, label="A (1,1)")
plt.scatter(*B, s=100, label="B (4,3)")

# Euclidean: straight line
plt.plot(
    [A[0], B[0]],
    [A[1], B[1]],
    linestyle="--",
    label=f"Euclidean = {euclidean:.2f}"
)

# Manhattan: horizontal + vertical
plt.plot(
    [A[0], B[0], B[0]],
    [A[1], A[1], B[1]],
    label=f"Manhattan = {manhattan:.2f}"
)

plt.xlim(0, 5)
plt.ylim(0, 4)

plt.xlabel("X")
plt.ylabel("Y")
plt.title("Euclidean vs Manhattan Distance")
plt.grid()
plt.legend()

plt.show()
```

```python
#KNN

import numpy as np

X = np.array([
    [2, 60],
    [3, 65],
    [7, 90],
    [8, 95],
    [6, 85]
])

y = np.array([
    "Fail",
    "Fail",
    "Pass",
    "Pass",
    "Pass"
])

new_student=np.array([6, 80])
distances=np.linalg.norm(X-new_student, axis=1)
print(distances)
```

```python
from sklearn.preprocessing import StandardScaler

scaler=StandardScaler()

X_scaler=scaler.fit_transform(X)

scaled_new_student=scaler.transform(
    new_student.reshape(1, -1)
)
```

```python
distances = np.linalg.norm(
    X_scaler - scaled_new_student,
    axis=1
)

print(distances)
```

```python
k=3
nearest_indices = np.argsort(distances)[:k]

print(nearest_indices)
```
