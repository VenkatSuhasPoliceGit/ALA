```python
import numpy as np

#create an input array using numoy
X=np.array([[1.0, 0.0],
           [0.0, 1.0],
           [1.0, 1.0]])

print(X)
print("Shape:", X.shape)
```

```python
# create weights9query)
wq=np.array([
    [1.0, 0.0],
    [0.0,1.0]
])

#create weights(key)
wk=np.array([
    [0.5, 0.0],
    [0.0, 0.5]
])

wv=np.array([
    [2.0, 0.0],
    [0.0, 2.0]
])

Q=X@wq
K=X@wk
V=X@wv

print(Q)
print(K)
print(V)
```

```python
# dimension of K
d_atten=K.shape[1]

scores=(Q@K.T)/np.sqrt(d_atten)
print(scores)
```

```python
mask = np.triu(np.ones((3, 3)), k=1).astype(bool)

print(mask)
```

```python
masked_scores = np.where(mask, -np.inf, scores)

print(masked_scores)
```

```python
#apply softmax
def softmax(x):
    exp_x = np.exp(x - np.max(x, axis=-1, keepdims=True))
    return exp_x / exp_x.sum(axis=-1, keepdims=True)

weights = softmax(masked_scores)
print(weights.round(3))
```
